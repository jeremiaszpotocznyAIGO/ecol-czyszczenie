```python

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

HEADERS = {
    "User-Agent": "Mozilla/5.0"
}

start_url = "https://ecol.eu/technologie-i-uslugi/uslugi-diagnostyczne/badania-wykonywane-w-laboratorium-ecol/"

def get_html(url):
    response = requests.get(url, headers=HEADERS, timeout=20)
    response.raise_for_status()
    return response.text


def extract_zastosowanie(url):
    html = get_html(url)
    soup = BeautifulSoup(html, "html.parser")

    for li in soup.find_all("li"):
        text = li.get_text(" ", strip=True)

        if "Zastosowanie:" in text:
            value = text.split("Zastosowanie:", 1)[1].strip()
            return [item.strip(" .;•") for item in value.split(",") if item.strip()]

        if "zastosowanie:" in text.lower():
            idx = text.lower().find("zastosowanie:")
            value = text[idx + len("zastosowanie:"):].strip()
            return [item.strip(" .;•") for item in value.split(",") if item.strip()]

    return []


html = get_html(start_url)
soup = BeautifulSoup(html, "html.parser")

results = []

for row in soup.select("div.table__row"):
    title_el = row.select_one("div.table__col.table__col--ornament h4.heading")
    link_el = row.select_one("a.table__link")

    if not title_el or not link_el:
        continue

    title = title_el.get_text(" ", strip=True)
    link = urljoin(start_url, link_el.get("href", "").strip())
    zastosowanie = extract_zastosowanie(link)

    results.append({
        "TITLE": title,
        "ZASTOSOWANIE": zastosowanie
    })

results
```

```python
import pandas as pd
import math
from collections import Counter

# 1) result columns only
result_cols = sorted({
    c
    for cols in group_category_to_cols.values()
    for c in cols
    if c in df.columns
})

binary_df = df[result_cols].notna().astype(int)

group_to_colset = {
    group: set(cols) & set(result_cols)
    for group, cols in group_category_to_cols.items()
}

# 2) weights: columns that belong to fewer groups get larger weight
col_group_counts = Counter()
for group, cols in group_to_colset.items():
    for c in cols:
        col_group_counts[c] += 1

n_groups = len(group_to_colset)
col_weights = {
    c: math.log((n_groups + 1) / (col_group_counts[c] + 1)) + 1
    for c in result_cols
}

def weighted_score(sample_set, group_set, col_weights):
    inter_cols = sample_set & group_set

    inter_w = sum(col_weights[c] for c in inter_cols)
    sample_w = sum(col_weights[c] for c in sample_set)
    group_w = sum(col_weights[c] for c in group_set)
    union_w = sample_w + group_w - inter_w

    precision = inter_w / sample_w if sample_w else 0.0
    recall = inter_w / group_w if group_w else 0.0
    iou = inter_w / union_w if union_w else 0.0

    if precision + recall > 0:
        f2 = (5 * precision * recall) / (4 * precision + recall)
    else:
        f2 = 0.0

    return {
        "weighted_intersection": inter_w,
        "weighted_precision": precision,
        "weighted_recall": recall,
        "weighted_iou": iou,
        "weighted_f2": f2,
        "n_matched_cols": len(inter_cols),
    }

specific_groups = [
    g for g in group_to_colset
    if g not in {"industrial_oil", "generic_oil", "other"}
]

fallback_groups = [g for g in ["industrial_oil", "generic_oil", "other"] if g in group_to_colset]

results = []

for idx, row in binary_df.iterrows():
    sample_set = set(row.index[row == 1])

    row_scores = []

    # first score specific groups
    for group in specific_groups:
        s = weighted_score(sample_set, group_to_colset[group], col_weights)
        s["group"] = group
        row_scores.append(s)

    row_scores_df = pd.DataFrame(row_scores).sort_values(
        ["weighted_f2", "n_matched_cols", "weighted_precision"],
        ascending=False
    )

    best_specific = row_scores_df.iloc[0].to_dict()

    # fallback if specific match is weak
    if best_specific["weighted_f2"] < 0.35 and fallback_groups:
        fallback_scores = []
        for group in fallback_groups:
            s = weighted_score(sample_set, group_to_colset[group], col_weights)
            s["group"] = group
            fallback_scores.append(s)

        fallback_df = pd.DataFrame(fallback_scores).sort_values(
            ["weighted_f2", "n_matched_cols", "weighted_precision"],
            ascending=False
        )
        best = fallback_df.iloc[0].to_dict()
    else:
        best = best_specific

    best["row_idx"] = idx
    results.append(best)

predicted_group_df = pd.DataFrame(results)
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import pairwise_distances
from sklearn.cluster import AgglomerativeClustering

# binary matrix
X = df[result_cols].notna().astype("uint8")

# exact pattern per row
pattern_id = X.astype(str).agg("".join, axis=1)

# one row per unique pattern
unique_patterns = (
    X.assign(pattern_id=pattern_id)
     .drop_duplicates("pattern_id")
     .set_index("pattern_id")
)

# keep only most frequent patterns to stay memory-safe
pattern_counts = pattern_id.value_counts()
top_n = 200  # adjust if needed
top_pattern_ids = pattern_counts.head(top_n).index

U = unique_patterns.loc[top_pattern_ids]

# boolean numpy array
U_bool = U.astype(bool).to_numpy()

# pairwise distances only on unique patterns
dist_matrix = pairwise_distances(U_bool, metric="jaccard")

n_clusters = 8
clustering = AgglomerativeClustering(
    n_clusters=n_clusters,
    metric="precomputed",
    linkage="average"
)

pattern_cluster_labels = clustering.fit_predict(dist_matrix)

pattern_cluster_df = pd.DataFrame({
    "pattern_id": top_pattern_ids,
    "cluster": pattern_cluster_labels,
    "count": pattern_counts.loc[top_pattern_ids].values
})

display(pattern_cluster_df.sort_values(["cluster", "count"], ascending=[True, False]))
```

```python
from itertools import combinations
from collections import Counter
import pandas as pd

tmp = binary_df.copy()
tmp["group"] = predicted_group_df["group"].values

group_pair_stats = []
 
all_rows = len(tmp)

for group in tmp["group"].dropna().unique():
    in_group = tmp[tmp["group"] == group].drop(columns="group")
    out_group = tmp[tmp["group"] != group].drop(columns="group")

    pair_in = Counter()
    pair_out = Counter()

    for _, row in in_group.iterrows():
        present = row.index[row == 1].tolist()
        for pair in combinations(sorted(present), 2):
            pair_in[pair] += 1

    for _, row in out_group.iterrows():
        present = row.index[row == 1].tolist()
        for pair in combinations(sorted(present), 2):
            pair_out[pair] += 1

    n_in = len(in_group)
    n_out = len(out_group)

    for pair, c_in in pair_in.items():
        c_out = pair_out.get(pair, 0)

        support_in = c_in / n_in if n_in else 0
        support_out = c_out / n_out if n_out else 0
        precision = c_in / (c_in + c_out) if (c_in + c_out) else 0
        lift = support_in / support_out if support_out > 0 else float("inf")

        group_pair_stats.append({
            "group": group,
            "col1": pair[0],
            "col2": pair[1],
            "count_in_group": c_in,
            "count_out_group": c_out,
            "support_in_group": support_in,
            "support_out_group": support_out,
            "precision_for_group": precision,
            "lift": lift,
        })

group_pair_df = pd.DataFrame(group_pair_stats)
```

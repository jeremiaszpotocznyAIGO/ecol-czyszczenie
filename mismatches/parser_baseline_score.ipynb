```python
import pandas as pd

df = pd.read_csv("all_rows_merged.csv")
df

parser_baseline_path = 'baseline_parsed_samples_all.json'

import json

with open(parser_baseline_path, "r", encoding="utf-8") as f:
    parser_baseline = json.load(f)

parser_baseline


def get_lab_with_interpretation(lab_number, parser_baseline, df):
    if lab_number not in parser_baseline:
        return None

    row = df.loc[df["Lab_Number"].astype(str) == str(lab_number)]

    if row.empty:
        return None

    row = row.iloc[0]  # first matching row

    result = dict(parser_baseline[lab_number])  # copy
    result["Overall_Interpretation"] = row["Overall_Interpretation"]
    result["Overall_Assessment"] = row["Overall_Assessment_pred_file"]

    return result


lab_data = get_lab_with_interpretation("P1304884", parser_baseline, df)
print(lab_data)
```
